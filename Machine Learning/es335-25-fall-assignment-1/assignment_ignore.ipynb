```python
from sklearn.datasets import make_classification
X, y = make_classification(
n_features=2, n_redundant=0, n_informative=2, random_state=1, n_clusters_per_class=2, class_sep=0.5)

# For plotting
import matplotlib.pyplot as plt
plt.scatter(X[:, 0], X[:, 1], c=y)
```

```python
# train test split
split_index = int(0.7 * len(X))
x_train, x_test, y_train, y_test = X[:split_index], X[split_index:], y[:split_index], y[split_index:]
```

```python
from tree.base import DecisionTree
clf = DecisionTree(max_depth=5,criterion='gini_index')
clf.fit(pd.DataFrame(x_train), pd.Series(y_train))
predictions = clf.predict(pd.DataFrame(x_test))
# make marker conditional on true label

plt.subplot(1, 2, 1)
plt.scatter(x_test[:, 0], x_test[:, 1], c=predictions, marker='o')
plt.title('Predictions')
plt.subplot(1, 2, 2)
plt.scatter(x_test[:, 0], x_test[:, 1], c=y_test, marker='x')
plt.title('True Labels')
plt.show()
```

```python
clf.plot()
```

```python
from sklearn.tree import DecisionTreeRegressor,DecisionTreeClassifier, plot_tree
reg = DecisionTreeClassifier(max_depth=5,criterion='gini')
reg.fit(pd.DataFrame(x_train), pd.Series(y_train))
predictions = reg.predict(pd.DataFrame(x_test))
plt.subplot(1, 2, 1)
plt.scatter(x_test[:, 0], x_test[:, 1], c=predictions, marker='o')
plt.title('Predictions')
plt.subplot(1, 2, 2)
plt.scatter(x_test[:, 0], x_test[:, 1], c=y_test, marker='x')
plt.title('True Labels')
plt.show()
```

```python
plot_tree(reg)
```

```python
url = './auto+mpg/auto-mpg.data'
data = pd.read_csv(url, delim_whitespace=True, header=None,
                 names=["mpg", "cylinders", "displacement", "horsepower", "weight",
                        "acceleration", "model year", "origin", "car name"])
```

```python
from tree.utils import one_hot_encoding
data = one_hot_encoding(data)
data
```

```python
data.drop(columns=['car name'], inplace=True)
data['horsepower'] = pd.to_numeric(data['horsepower'], errors='coerce')
data.dropna(inplace=True)
data.reset_index(drop=True, inplace=True)
x1=data.drop(columns=['mpg']).copy()
x2=data.drop(columns=['origin']).copy()
y1=data['mpg'].copy()
y2=data['origin'].copy()
print(x1.shape, x2.shape, y1.shape, y2.shape)
```

```python
## regression tree
from tree.base import DecisionTree
from matplotlib import pyplot as plt
cus_reg = DecisionTree(max_depth=3,criterion='mse')
cus_reg.fit(x1, y1)
cus_reg.plot()
predictions = cus_reg.predict(x1)
plt.scatter(y1, predictions)
plt.xlabel("True Values")
plt.ylabel("Predictions")
plt.axis('equal')

from metrics import rmse,mae
print("root mean square error ",rmse(y1, predictions), "mean absolute error ",mae(y1, predictions))
```

```python
## regression tree using sklearn
from sklearn.tree import DecisionTreeRegressor, plot_tree
from sklearn.metrics import mean_squared_error, mean_absolute_error
reg = DecisionTreeRegressor(max_depth=3,criterion='squared_error')
reg.fit(x1, y1)
predictions = reg.predict(x1)
plt.scatter(y1, predictions)
plt.xlabel("True Values")
plt.ylabel("Predictions")
plt.axis('equal')
plt.title("Sklearn Decision Tree Regressor")
plt.show()
plot_tree(reg)
print("root mean square error: ",mean_squared_error(y1, predictions)**0.5, "Mean absolute error: ",mean_absolute_error(y1, predictions))
```

```python
# Classification tree
from metrics import accuracy, precision, recall
#confusion matrix
from sklearn.metrics import confusion_matrix
clf = DecisionTree(max_depth=5,criterion='gini_index')
clf.fit(x2, y2)
predictions = clf.predict(x2)
print("Accuracy ",accuracy(y2, predictions))
clf.plot()
confusion_matrix(y2, predictions)
```

```python
# classification tree using sklearn
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.metrics import accuracy_score, precision_score, recall_score
reg = DecisionTreeClassifier(max_depth=5,criterion='gini')
reg.fit(x2, y2)
predictions = reg.predict(x2)
plot_tree(reg)
print("Accuracy ",accuracy_score(y2, predictions), "Precision ",precision_score(y2, predictions, average='weighted'), "Recall ",recall_score(y2, predictions, average='weighted'))
print(confusion_matrix(y2, predictions))
```
